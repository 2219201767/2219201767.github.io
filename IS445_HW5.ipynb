```python

import pandas as pd
import altair as alt
url = "https://raw.githubusercontent.com/UIUC-iSchool-DataViz/is445_data/main/bfro_reports_fall2022.csv"

df = pd.read_csv(url)

print("Dataset shape:", df.shape)
print("Column names:", df.columns.tolist())

df.head()
```

```python
state_counts = (
    df.dropna(subset=["state"])
      .groupby("state")
      .size()
      .reset_index(name="count")
      .sort_values("count", ascending=False)
      .head(15)
)

selection = alt.selection_point(fields=["state"])

chart1 = (
    alt.Chart(state_counts)
    .mark_bar()
    .encode(
        x=alt.X("count:Q", title="Number of Reports"),
        y=alt.Y("state:N", sort="-x", title="State"),
        color=alt.condition(
            selection,
            alt.value("steelblue"),
            alt.value("lightgray")
        ),
        tooltip=[
            alt.Tooltip("state:N", title="State"),
            alt.Tooltip("count:Q", title="Reports")
        ]
    )
    .add_params(selection)
    .properties(
        title="Top 15 States by Bigfoot Sightings",
        width=550,
        height=400
    )
)

chart1
```

```python
season_order = ["Spring", "Summer", "Fall", "Winter"]

season_counts = (
    df[df["season"].isin(season_order)]
    .groupby("season")
    .size()
    .reset_index(name="count")
)
chart2 = (
    alt.Chart(season_counts)
    .mark_bar()
    .encode(
        x=alt.X(
            "season:N",
            sort=season_order,
            title="Season"
        ),
        y=alt.Y(
            "count:Q",
            title="Number of Reports"
        ),
        color=alt.Color(
            "season:N",
            scale=alt.Scale(
                domain=season_order,
                range=["#66A61E", "#E6AB02", "#D95F02", "#7570B3"]
            ),
            legend=None
        ),
        tooltip=[
            alt.Tooltip("season:N", title="Season"),
            alt.Tooltip("count:Q", title="Reports")
        ]
    )
    .properties(
        title="Bigfoot Sightings by Season",
        width=500,
        height=350
    )
)

chart2

```

```python
chart1.save("chart1.html")
chart2.save("chart2.html")
```
